## 1. What is BERT?

BERT is an **Encoder-based Transformer** developed by Google.
- **Bidirectional**: Unlike older models that read text left-to-right (or right-to-left), BERT reads the *entire sequence of words simultaneously*.
- **Encoder**: Focuses on extracting rich contextual representations from text.

### Practical Application: Sentiment Classification
Let's see how BERT processes a sentence like:
`"The movie was absolutely fantastic!"`
and classifies whether the sentiment is **POSITIVE** or **NEGATIVE**.



---
## 2. Code Example: First Sentiment Analysis Pipeline

Below is complete runnable Python code using Hugging Face `transformers`.



```python
from transformers import pipeline

# Load a pre-trained sentiment analysis pipeline powered by DistilBERT (a lightweight BERT model)
classifier = pipeline("sentiment-analysis")

# Input text
text = "The movie was absolutely fantastic!"

# Run classification
result = classifier(text)

print("Input Text:", text)
print("Output Result:", result)

```

---
## 💡 Line-by-Line Code Breakdown

### ❓ 1. What are we doing?
We are taking an input sentence (`"The movie was absolutely fantastic!"`) and determining whether the emotional sentiment expressed in the sentence is positive or negative, along with a confidence score.

### ❓ 2. Why are we doing it?
Sentiment analysis is used in real-world applications such as analyzing customer product reviews, monitoring social media brand feedback, and filtering support tickets automatically.

### ❓ 3. How does the code work?
- `from transformers import pipeline`: Imports the `pipeline` helper function from the Hugging Face library.
- `classifier = pipeline("sentiment-analysis")`: Automatically downloads a pre-trained sentiment analysis model (by default, `distilbert-base-uncased-finetuned-sst-2-english`) and its corresponding tokenizer.
- `result = classifier(text)`: Passes our text through the tokenizer and model, returning a Python list containing a dictionary with `label` (`POSITIVE` / `NEGATIVE`) and `score` (confidence probability between 0 and 1).
