## 1. Summarization in Modern NLP

Unlike traditional extractive summarization (which picks out existing sentences), modern Encoder-Decoder models (like BART or T5) produce **abstractive summarization**—rewriting and compressing key ideas into new, concise sentences.

---
## 2. Code Example: Summarization Pipeline



```python
from transformers import pipeline

# Load pre-trained summarization pipeline
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

# Input passage
text = """
Artificial intelligence is becoming an important part of many industries. 
Companies use AI for customer service, recommendation systems, document processing, 
fraud detection, and many other applications. Recent advances in Transformer 
models have made it easier to build applications that understand and generate 
natural language without requiring custom trained models for every individual task.
"""

# Generate summary
summary = summarizer(
    text,
    max_length=45,
    min_length=15,
    do_sample=False
)

print("Original Text Length:", len(text.split()), "words")
print("\nGenerated Summary:")
print(summary[0]['summary_text'])

```

---
## 💡 Line-by-Line Code Breakdown

### ❓ 1. What are we doing?
We condense a multi-sentence paragraph into a 1-2 sentence core summary.

### ❓ 2. Why are we doing it?
Summarization powers executive briefing tools, news feed digests, email summaries, and document processing systems.

### ❓ 3. Parameter Controls Explained
- `max_length=45`: Sets upper limit on generated summary tokens.
- `min_length=15`: Forces the summary to contain sufficient detail.
- `do_sample=False`: Uses greedy/beam search for concise, factual, deterministic summaries.
