## 📍 Use Case 1: Customer Review Summarizer



```python
from transformers import pipeline

summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

review = """
The wireless noise-canceling headphones exceeded all my expectations! The audio quality is crisp, 
the battery lasts over 30 hours, and pairing with my phone was completely seamless. Highly recommended 
for travelers and remote workers who need quiet focus.
"""

summary = summarizer(review, max_length=30, min_length=10, do_sample=False)
print("Customer Review Summary:", summary[0]['summary_text'])

```

---
## 📍 Use Case 2: Meeting Notes Summarizer



```python
meeting_notes = """
Project Alignment Meeting: Sarah announced launch set for Friday. David completed database caching 
which fixed API latency. Elena presented user feedback requesting dark mode. Action items: David 
will run load tests by Thursday; Elena will send UI mockups by Wednesday.
"""

summary_notes = summarizer(meeting_notes, max_length=40, min_length=15, do_sample=False)
print("Meeting Highlights:", summary_notes[0]['summary_text'])

```

## 📍 Use Case 3: Email Assistant



```python
email_body = """
Hi Team, Please be reminded that third-party authentication tokens expire on August 15th. 
All developers must update their API access keys in the cloud portal before 5:00 PM EST 
to prevent service disruption.
"""

email_summary = summarizer(email_body, max_length=30, min_length=10, do_sample=False)
print("Email Summary:", email_summary[0]['summary_text'])

```

## 📍 Use Case 4: Topic Explainer



```python
generator = pipeline("text-generation", model="distilgpt2")

prompt_explain = "Topic: Machine Learning\nAudience: Beginner Student\nExplanation: Machine learning is"
explanation = generator(prompt_explain, max_length=60, do_sample=True, temperature=0.7)

print("Generated Explanation:")
print(explanation[0]['generated_text'])

```

## 📍 Use Case 5: Prompt Comparison Experiment
Compare three prompts with varying levels of specificity on the exact same model!



```python
prompt_a = "Explain machine learning."
prompt_b = "Explain machine learning to a beginner."
prompt_c = "Explain machine learning to a beginner student using one simple real-world example:"

print("Prompt A output:", generator(prompt_a, max_length=45, do_sample=True)[0]['generated_text'])
print("\nPrompt B output:", generator(prompt_b, max_length=45, do_sample=True)[0]['generated_text'])
print("\nPrompt C output:", generator(prompt_c, max_length=45, do_sample=True)[0]['generated_text'])

```
