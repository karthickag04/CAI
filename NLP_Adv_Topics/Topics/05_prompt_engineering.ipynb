## 1. What is a Prompt?

> A **prompt** is the natural language instruction or input text provided to a language model to guide its output.

Prompting requires no model retraining or parameter changes. By improving the instruction, you dramatically improve output utility!

---
## 2. Prompt Progression Strategy

Let's examine how prompt clarity changes model results:

### Level 1: Basic Prompt
```text
Explain artificial intelligence.
```

### Level 2: Specific Prompt
```text
Explain artificial intelligence in simple language for a college student.
```

### Level 3: Structured Prompt
```text
Explain artificial intelligence for a beginner.

Include:
1. Definition
2. One real-world example
3. One advantage
4. One limitation

Keep the answer under 100 words.
```



````python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")

prompt_basic = "Artificial Intelligence is defined as"
prompt_structured = "Artificial Intelligence for beginners: 1. Definition: AI is"

out1 = generator(prompt_basic, max_length=50, do_sample=True, temperature=0.7)
out2 = generator(prompt_structured, max_length=60, do_sample=True, temperature=0.7)

print("=== Basic Prompt Output ===")
print(out1[0]['generated_text'])

print("\n=== Structured Prompt Output ===")
print(out2[0]['generated_text'])

````

---
## 💡 Key Takeaways for Prompt Engineering
1. **Clear Instructions**: State explicitly what you want.
2. **Assign a Role**: "Act as a computer science instructor..."
3. **Specify Constraints**: Word counts, bullet points, target audience.
4. **Provide Examples (Few-Shot)**: Show sample input/output pairs.
