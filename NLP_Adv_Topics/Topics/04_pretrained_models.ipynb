
## 2. Code Example: Text Generation with `distilgpt2`



````python
from transformers import pipeline

# Load a specific pre-trained decoder model: DistilGPT-2
generator = pipeline("text-generation", model="distilgpt2")

# Define prompt
prompt = "Artificial intelligence will transform the future of"

# Generate text
outputs = generator(
    prompt,
    max_length=50,
    num_return_sequences=2,
    do_sample=True,
    temperature=0.7
)

print("Prompt:", prompt)
print("-" * 50)
for i, out in enumerate(outputs, 1):
    print(f"Completion {i}:\n{out['generated_text']}\n")

````

---
## 💡 Line-by-Line Code Breakdown

### ❓ 1. What are we doing?
Given an initial prompt phrase, we ask a pre-trained autoregressive Transformer (`distilgpt2`) to generate continuing text completions.

### ❓ 2. Why are we doing it?
Text generation powers AI writing assistants, auto-completion, story generation, and code generation tools.

### ❓ 3. Key Parameters Explained
- `model="distilgpt2"`: Explicitly chooses a lightweight version of GPT-2 from Hugging Face Model Hub.
- `max_length=50`: Sets maximum total token count (prompt + generated tokens).
- `num_return_sequences=2`: Asks the model to generate 2 distinct completions.
- `do_sample=True`: Enables sampling mode so completions vary creatively.
- `temperature=0.7`: Controls randomness (lower = predictable, higher = creative).

## 🧪 4. What can I change? (Student Experiment)
Try changing `prompt` and experimenting with `temperature` (try `0.2` vs `0.9`)!



````python
test_prompt = "In ten years, autonomous vehicles will"
sample_low_temp = generator(test_prompt, max_length=40, temperature=0.2, do_sample=True)
sample_high_temp = generator(test_prompt, max_length=40, temperature=0.9, do_sample=True)

print("--- Low Temperature (0.2) ---")
print(sample_low_temp[0]['generated_text'])

print("\n--- High Temperature (0.9) ---")
print(sample_high_temp[0]['generated_text'])

````
