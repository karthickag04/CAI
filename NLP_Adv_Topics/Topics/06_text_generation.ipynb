## 🛠️ Practical Use Case: Course Description Generator

Imagine an online learning platform that needs to generate attractive course descriptions automatically based on course metadata.

### Scenario:
- **Course Title**: Python Programming Fundamentals
- **Audience**: Beginners
- **Duration**: 6 weeks
- **Key Topics**: Functions, loops, file handling, basic projects



```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")

# Contextual prompt
prompt = (
    "Course Title: Python Programming Fundamentals\n"
    "Target Audience: College Beginners\n"
    "Duration: 6 Weeks\n"
    "Description: Learn Python programming from scratch! In this course, you will cover"
)

result = generator(prompt, max_length=100, do_sample=True, temperature=0.7)
print("Generated Course Description:\n")
print(result[0]['generated_text'])

```

---
## 💡 Line-by-Line Code Breakdown

### ❓ 1. What are we doing?
We construct a structured context prompt containing course details and ask the model to generate a cohesive course overview description.

### ❓ 2. Why are we doing it?
Automating content generation reduces manual writing effort for marketing, educators, and product catalog creators.

### ❓ 3. How does the code work?
We pass key metadata (Title, Target Audience, Duration) at the beginning of the prompt string. The Transformer decoder uses this prefix context to constrain its generation tokens.
