# Notebook 09: Mini Project — Student Notes Assistant

## 🎯 Learning Objectives
Build a complete mini-project that takes raw student study notes and processes them into a structured output containing:
1. **Short Summary**
2. **Key Highlights**
3. **Important Term Extraction**

---
## 🏗️ Project Architecture

```text
Raw Student Notes ──> Summarization Pipeline ──> Summary & Highlights
                  ──> NER / Term Pipeline    ──> Important Key Terms
```



````python
from transformers import pipeline

# Load pre-trained pipelines
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")
ner_tagger = pipeline("ner", aggregation_strategy="simple")

def analyze_student_notes(notes_text):
    print("=" * 60)
    print(" 📖 STUDENT NOTES ASSISTANT REPORT")
    print("=" * 60)
    
    # 1. Summary
    summary_res = summarizer(notes_text, max_length=50, min_length=15, do_sample=False)
    summary_text = summary_res[0]['summary_text']
    
    # 2. Extract Key Entities/Terms
    entities = ner_tagger(notes_text)
    extracted_terms = list(set([e['word'] for e in entities]))
    
    print("\n📌 1. SHORT SUMMARY:")
    print(summary_text)
    
    print("\n🏷️ 2. EXTRACTED KEY TERMS / ENTITIES:")
    if extracted_terms:
        for term in extracted_terms:
            print(f" - {term}")
    else:
        print(" - No specific named entities found.")
        
    print("\n💡 3. ACTIONABLE STUDY ADVICE:")
    print("Review the summary above and test yourself on the extracted terms!")
    print("=" * 60)

# Sample Student Notes Input
raw_notes = """
Machine learning is a branch of artificial intelligence focused on building applications 
that learn from data. Supervised learning uses labeled training data, whereas unsupervised 
learning finds patterns in unlabeled data. Deep learning relies on artificial neural networks. 
Google and Microsoft use these techniques extensively in search engines.
"""

# Run Assistant
analyze_student_notes(raw_notes)

````
