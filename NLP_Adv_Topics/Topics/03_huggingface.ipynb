## 1. What is Hugging Face?

Hugging Face provides an open-source library (`transformers`) and platform (**Model Hub**) containing tens of thousands of pre-trained models.

```text
Hugging Face Pipeline Architecture:

Input Raw Text ──> Tokenizer (Text → IDs) ──> Transformer Model (IDs → Vector Output) ──> Post-Processing (Output)
```

- **Tokenizer**: Converts raw text into numerical token IDs that Transformer models understand.
- **Model**: The neural network containing learned weights.
- **Pipeline**: High-level wrapper that automatically connects Tokenizer + Model + Post-processing in 1 line of code.

---
## 2. Using Pipelines for Named Entity Recognition (NER)

Students already know what NER is (identifying entities like Persons, Organizations, Locations).
Here, we demonstrate how to perform NER using a powerful pre-trained Transformer model with very little code!



````python
from transformers import pipeline

# Initialize NER pipeline with simple aggregation (combines subword tokens into full entity names)
ner_pipeline = pipeline("ner", aggregation_strategy="simple")

# Input text
text = "Apple CEO Tim Cook visited India last week to open a new store in Mumbai."

# Run NER
entities = ner_pipeline(text)

print("Input Text:", text)
print("\nDetected Entities:")
for entity in entities:
    print(f" - Entity: '{entity['word']}' | Group: {entity['entity_group']} | Score: {entity['score']:.4f}")

````

---
## 💡 Line-by-Line Code Breakdown

### ❓ 1. What are we doing?
We are using a Transformer model to extract named entities (person names, company names, location names) from a news sentence.

### ❓ 2. Why are we doing it?
Instead of building complex rule-based taggers or training models from scratch, we leverage a pre-trained Transformer that already understands entity contexts globally.

### ❓ 3. How does the code work?
- `pipeline("ner", aggregation_strategy="simple")`: Loads a pre-trained NER model. Setting `aggregation_strategy="simple"` ensures multi-token words (like `"Tim Cook"`) are grouped together logically rather than split into subword tags.
- `ner_pipeline(text)`: Returns a list of dictionaries with detected `entity_group`, `word`, and `score`.
